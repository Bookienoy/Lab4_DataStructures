```python
# Student Performance Data Analyzer
last_name = "Anonuevo"
seed_num = 9
favorite_artist = "Taylor swift"
modifier = len(favorite_artist) # 15 characters

students = [f"{last_name}_S1", f"{last_name}_S2", f"{last_name}_S3", f"{last_name}_S4", f"{last_name}_S5"]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

generated_records = [
    {"student": students[0], "subject": subjects[0], "score": seed_num * 12 + modifier}, # 87
    {"student": students[1], "subject": subjects[1], "score": seed_num * 10 + modifier}, # 75
    {"student": students[2], "subject": subjects[2], "score": seed_num * 8 + modifier},  # 63
    {"student": students[0], "subject": subjects[0], "score": seed_num * 12 + modifier}, # Repeated record
    {"student": students[3], "subject": subjects[3], "score": None} # Incomplete record (missing score)
]
print("1. Generated Records:", generated_records)

organized_data = {}
for r in generated_records:
    stu = r["student"]
    sub = r["subject"]
    if stu not in organized_data:
        organized_data[stu] = {}
    organized_data[stu][sub] = r["score"]
print("2. Organized Data:", organized_data)

print("3. Repeated/Incomplete: 1 duplicate found (Duario_S1 PROGRAMMING) and 1 incomplete record (Duario_S4 PHYSICS with None score).")

def get_category(score):
    if score is None: return "Incomplete"
    elif score >= 85: return "Excellent"
    elif score >= 70: return "Good"
    else: return "Passing"

categories_result = [(r["student"], r["subject"], get_category(r["score"])) for r in generated_records]
print("4. Performance Categories:", categories_result)

valid_scores = [r["score"] for r in generated_records if r["score"] is not None]
avg_score = sum(valid_scores) / len(valid_scores)
print(f"5. Summaries - Average Valid Score: {avg_score}")

ordered_results = sorted([r for r in generated_records if r["score"] is not None], key=lambda x: x["score"], reverse=True)
print("6. Ordered Results (Descending):", ordered_results)

print("7. Final Student Performance Report: Successfully compiled, filtered, and analyzed student records.")
```
