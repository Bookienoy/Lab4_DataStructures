```python
LAST_NAME = "Anonuevo"
SEED_NUM = 9
FAVORITE_ARTIST = "taylor swift"

# --- TASK 1: GENERATE TEST DATA ---
# 1. Test Sources: LAST_NAME with labels T1 to T5
test_sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]

# 2. Measurement Categories
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

# 3. Measurement Values derived from SEED_NUM and character positions of FAVORITE_ARTIST
# Generating values using ASCII character codes adjusted by seed
generated_values = [ord(char) * SEED_NUM % 100 + 10.5 for char in FAVORITE_ARTIST]

# Construct raw test records: list of tuples (Source, Category, Value)
raw_records = [
    (test_sources[0], categories[0], generated_values[0]),  # TEMPERATURE
    (test_sources[1], categories[1], generated_values[1]),  # VOLTAGE
    (test_sources[2], categories[2], generated_values[2]),  # CURRENT
    (test_sources[3], categories[3], generated_values[3]),  # PRESSURE
    (test_sources[4], categories[4], generated_values[4]),  # SPEED
    # Additional data: multiple measurements under categories (TEMPERATURE & VOLTAGE)
    (test_sources[0], categories[0], generated_values[5]),  # TEMPERATURE
    (test_sources[1], categories[1], generated_values[6]),  # VOLTAGE
    # Repeated record requirement
    (test_sources[0], categories[0], generated_values[0]),  # REPEATED RECORD
]

print("=== 1. GENERATED TEST DATA (RAW RECORDS) ===")
for r in raw_records:
    print(f"Source: {r[0]:<12} | Category: {r[1]:<12} | Value: {r[2]:.2f}")
print("-" * 60)
```
